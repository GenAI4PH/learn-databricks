```python
from pyspark.sql.functions import sum, col

df_sales = spark.read.table("samples.bakehouse.sales_transactions")
df_customer = spark.read.table("samples.bakehouse.sales_customers")

df_total_sales = df_sales.join(df_customer, "customerID").groupBy("first_name","last_name").agg(
    sum("totalPrice").alias("total_sales"),
    sum(col("quantity") * col("unitPrice")).alias("total_quantity")
)



```
